## **The Unethical Use of Health Data - The Case of Henrietta Lacks**

One of the best-known cases of the unethical use of biological information from a patient in health research is that of Henrietta Lacks, who was a Black woman diagnosed with cervical cancer in 1951. During her treatment for cervical cancer at Johns Hopkins Hospital, doctors took samples from the tumour located in her cervix without her knowledge and without obtaining her consent. The samples they obtained from her were of living cancer cells that continued to grow (later named as HeLa cells) and have now become a source of research used for many of the significant advancements in modern medicine today.

The ethical shortcomings in this case are many. Henrietta was never informed that her cells would be removed or were being removed, nor was she given the opportunity to decide if she would allow doctors to take her cells from her without her knowledge. For many years after Henrietta's death (she died in 1951 from her cervical cancer), her family was not aware that her cells had been taken, and her cells had been shipped, sold and used, not only for the continuous development of the HeLa cells into science, but also used and marketed for profit. The actions of the doctors from Johns Hopkins Hospital violated the principles of informed consent, autonomy, and respect for persons and were especially significant in the context of the historically documented exploitation of individuals and groups that represent marginalized populations in health care research.

The personal and social impact of the unethical behaviour surrounding the creation of HeLa cells was multi-faceted, while the doctors from Johns Hopkins Hospital may have advanced the knowledge of science via scientific advances gained through use of HeLa cells (e.g., vaccines and cancer treatment), the Lacks family experienced feelings of sadness and mistrust towards the health care and the health care system. In addition, the Henrietta Lacks case illuminated the systemic disparities that exist in how the rights of patients are respected.

Skloot, R. (2010). The immortal life of Henrietta Lacks. Crown Publishing Group.

## **Bias in Health Data Science - Racial Bias in Healthcare Algorithms**

The author's research examines how to address racial bias in health data science through the perspective of an algorithm developed to predict patients who are likely to require more medical care. They highlight a case study conducted in 2019 examining a popular algorithm used across the U.S. healthcare system to identify patients who would benefit from additional medical care. The findings from this study reveal that the algorithm disproportionately underestimated the health care needs of Black patients.

The basis for the algorithm's flawed decisions stemmed from using healthcare spending as a proxy for illness. In the case of Black patients, there have historically been structural inequalities placed upon them resulting in lower healthcare spending compared to their equally or more serious health needs. Using healthcare spending to estimate the health of Black patients led to incorrect conclusions that Black patients were healthy and caused fewer Black patients to be entered or taken care of by the programs aimed at helping those with greater health care needs.

The authors provide insight into the 2019 study's results, including the impact of this racial bias on patients receiving timely preventive services and follow-up care. This study also further reinforces that the need for effective and responsive health care systems is essential to eliminating long-term racial disparities in systemic healthcare outcomes, resulting from biased information and assumptions about the health of Black patients.

There are multiple avenues for preventing similar bias from occurring again using, utilizing direct health indicator measures versus spending data when developing healthcare algorithms, evaluating the performance of algorithms with a diverse sample of patients prior to implementing an algorithm, as well as performing a bias audit of the algorithm prior to implementation. Involving ethicists as well as diverse stakeholders during the development of healthcare algorithms can assist in reducing the number of racially discriminatory outcomes resulting from the use of biased health data science.

## **Anonymization of Health Data - Age Shifting Technique**

In order to protect patients' privacy, we need to anonymize health data, and still provide useful research information from these data. One of the most common forms of anonymization is to use an age shifting technique. In this technique, we adjust the patients' ages by a set number of years for all patients in a small group of patients. For example, if we have a group of 5 patients who are all 40 years old, we could add three years to each of those patients' ages, or subtract three years from each of their ages (i.e. 43 and 37 respectively). This method provides a means to mask patient identities while still maintaining the relative age of patients in comparison to one another.

The age shifting technique reduces the chances of re-identification through the elimination of exact matches with external data. All the patients in the group will have a slightly different age but will have been altered consistently from one another making it easy to analyze the age-related risk trends without compromising the integrity of an individuals' confidential data.

The main advantage of this technique for protecting patient confidentiality and anonymity is its relative simplicity and effectiveness at achieving the desired level of privacy protection while still providing a basis to analyze records for the purpose of identifying potential trends in health information. However, use of this method alone does not guarantee patient confidentiality and anonymity and should be performed with caution and in conjunction with other anonymization methods.

It is possible to enhance the effectiveness of age shifting through the use of established clustering methods such as range groupings and/or k-anonymity. Ongoing monitoring and frequent analysis of the ability to re-identify patients continue to be critical as additional external data sets become accessible.

## **Pro-Form Consent Document for Health Data Collection**

The NIH's standardized informed consent template provides an example of a pro-forma consent document used to collect health-related data. The template sets forth the purpose of the study, data to be collected, use of the data, possible risks and benefits of participating, and the right of the participant to withdraw from the study at any time.

The pro forma consent document serves as a means to obtain informed consent by presenting information in a concise and understandable manner. It protects the autonomy of the participants and creates transparency between the researcher and the participant. The need for pro forma consent documents become more critical in health-related data research due to the sensitive nature of personal data that will be collected.

Proforma documents provide a means to promote ethical consistency across studies by standardizing the procedures for obtaining consent and reducing the likelihood of misuse or misinterpretation of a study's data.

**References (APA):**
National Institutes of Health. (2023). Informed consent guidelines.

## **The George Washington University Office of Human Research**

The Office for Human Research at The George Washington University (GWU) provides support for research involving humans at GWU by overseeing that the research complies with federal regulations and is conducted ethically. They also work closely with Institutional Review Boards (IRBs) reviewing protocols, informed consent processes, and data security plans for studies conducted at GWU.

The main responsibility of the Office for Human Research is to protect the rights, privacy, and welfare of research participants and to ensure that researchers and studies conducted at GWU follow federal regulations including the Common Rule. Additionally, they provide education and support to researchers on conducting ethical research.

Through its oversight of the conduct of research, training of researchers, and monitoring of researchers, the GW Office for Human Research plays an integral role in maintaining public trust in academic research by preventing ethical violations.
