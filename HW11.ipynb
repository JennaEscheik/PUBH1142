Create an array of 1000 bootstrapped sample means (sample size = 40), assign it to `bootstrapped_means`, and sort it.

```python
numpy.random.seed(12)

bootstrapped_means = [
    numpy.random.choice(sample, 40, replace=True).mean()
    for _ in range(1000)
]

bootstrapped_means = numpy.sort(bootstrapped_means)
```
