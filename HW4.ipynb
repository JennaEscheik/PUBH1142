# **HW 4 | Spreadsheets and Databases**

1. **What is the difference between a spreadsheet and a database?**

A spreadsheet is a tool used to organize and analyze small datasets in a simple table format with rows and columns, while a database is designed to manage large, complex datasets using multiple related tables with built-in features for data integrity, security, and advanced querying.

2. **What are the advantages and disadvantages of using a spreadsheet for health data management?**

The advantages of using a spreadsheet for health data management include ease of use, quick data entry, built-in formulas, and simple visualization tools, while the disadvantages include limited scalability, weaker data integrity controls, limited multi-user access, and difficulty managing complex relationships.

3. **What are the advantages and disadvantages of using a database for health data management?**

The advantages of using a database for health data management include the ability to handle large datasets, maintain data integrity, support multiple users, and manage relationships between tables, while the disadvantages include greater complexity, the need for SQL knowledge, and more time required for setup and maintenance.

4. **What are some examples of tasks that can be performed using a spreadsheet?**

Examples of tasks that can be performed using a spreadsheet include entering patient data, calculating averages or totals, sorting and filtering data, creating charts, and using pivot tables to summarize information.

5. **What are some examples of tasks that can be performed using a database?**

Examples of tasks that can be performed using a database include managing electronic health records, linking patient demographics to lab results, running SQL queries for reports, controlling user access, and handling large-scale clinical trial data.

6. **What are some examples of tasks that can be performed using both a spreadsheet and a database?**

Both spreadsheets and databases can be used to store data, sort and filter records, generate reports, perform basic analyses, and import or export data in formats such as CSV.

7. **What are some examples of tasks that can be performed using neither a spreadsheet nor a database?**

Examples of tasks that cannot be performed using a spreadsheet or a database include performing surgery, conducting a physical medical examination, or directly administering medical treatment.

8. **What are some examples of tasks that can be performed using a spreadsheet but not a database?**

Tasks that can be performed using a spreadsheet but not easily in a database include applying detailed cell formatting, creating quick visual charts directly from selected cells, and performing simple ad hoc calculations without database queries.

9. **What are some examples of tasks that can be performed using a database but not a spreadsheet?**

Tasks that can be performed using a database but not effectively in a spreadsheet include managing millions of records efficiently, enforcing referential integrity between tables, handling simultaneous multi-user access, and executing complex relational queries using SQL.

10. **What are some examples of tasks that can be performed using both a spreadsheet and a database?**

Both spreadsheets and databases can be used to organize data, summarize information, filter datasets based on criteria, and generate basic analytical reports.
