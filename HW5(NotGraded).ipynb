### What is the purpose of an Integrated Development Environment (IDE)?

An IDE is a software environment that combines tools for writing, running, debugging, and managing code in one place (editor, debugger, terminal/build tools, version control integration).

### Which popular Python library is specifically designed for numerical and scientific computing?

### What is the purpose of the pandas library in Python?

To manipulate and analyze structured/tabular data using objects like DataFrames (cleaning, filtering, grouping, merging, etc.).

### Name the deep learning framework developed by Google.

TensorFlow

### Which Python library is commonly used for data visualization?

Matplotlib

### What is the main purpose of the stats module in the SciPy library?

To perform statistical analysis: distributions, descriptive stats, hypothesis tests, and related statistical functions.
